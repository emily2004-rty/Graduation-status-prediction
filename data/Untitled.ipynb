```python
#dropping columns not needed for prediction
data_model=data.drop(['Timestamp', 'ID No.','Motivation',
        'day', 'Hour'], axis=1)
```

```python
#identifying the x and y 
y=data_model["Graduated"]
x=data_model.drop("Graduated",axis=1)
```

```python
#importing necessary libraried  for predicting model
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.metrics import confusion_matrix,classification_report
from sklearn.preprocessing import OneHotEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.compose import ColumnTransformer
```

```python
#spiltting the y and x values to train and test
x_test,x_train,y_test,y_train=train_test_split(x,y, test_size=0.2,random_state=42)
```

```python
data_model.columns
```

```python
#identifying numericl and categorical columns
numerical=["Total score"]
categorical=["Gender","Age range","Hear_About_us","Track","Aim","Learning_Experience","Hours_commit","Skill level","Test_Completion","Country"]
```

```python
#performing standardrization to numerical values and one hot encoding to categorical columns
preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numerical),
        ("cat",OneHotEncoder(handle_unknown="ignore"), categorical)
    ],
    remainder="drop"
)
```

```python
#fitting the logisitic regression model
pipeline=Pipeline([
    ("prep",preprocessor),
    ("clf",LogisticRegression(class_weight="balanced",max_iter=1000))
])
```

```python
#fitting the x,y train into  the pipeline
pipeline.fit(x_train,y_train)
```

```python
y_pred =pipeline.predict(x_test)
```

```python
print(classification_report(y_test,y_pred))
```

```python
pipeline.score(x_test,y_test)*100
```

```python
import joblib
joblib.dump(pipeline,"model.pkl")
```

```python
#traing with randomforest classifier
from sklearn.ensemble import RandomForestClassifier
rfc=Pipeline([
    ("prep",preprocessor),
    ("classifier",RandomForestClassifier())
])
```

```python
rfc.fit(x_train,y_train)
```

```python
pred =rfc.predict(x_test)
```

```python
rfc.score(x_test,y_test)*100
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numerical),
        ("cat",OneHotEncoder(handle_unknown="ignore"), categorical)
    ],
    remainder="drop"
)
```

```python
kns= Pipeline([
    ('prep', preprocessor),
    ('knn', KNeighborsClassifier())
])
```

```python
param_grid = {
    'knn__n_neighbors': [3, 5, 7],
    'knn__weights': ['uniform', 'distance'],
    'knn__metric': ['euclidean', 'manhattan']
}
```

```python
# Set up GridSearchCV
grid_search = GridSearchCV(
    estimator=kns,
    param_grid=param_grid,
    cv=3,
    scoring='accuracy',
    n_jobs=-1
)
```

```python
grid_search.fit(x_train, y_train)
```

```python
print(" Best Parameters:", grid_search.best_params_)
print("Best CV Accuracy:", grid_search.best_score_)
```

```python
# Test accuracy
best_model = grid_search.best_estimator_
ypred = best_model.predict(x_test)
```

```python
best_model.score(x_test,y_test)*100
```

```python
print(classification_report(y_test,ypred))
```
