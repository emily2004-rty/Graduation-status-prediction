```python
#dropping columns not needding for prediction
data_model=data.drop(['Timestamp', 'ID No.','Motivation',
        'day', 'Hour'], axis=1)
```

```python
#identifying the x and y columns
y=data_model["Graduated"]
x=data_model.drop("Graduated",axis=1)
```

```python
#importing necessary libraried 32 for predicting model
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.metrics import confusion_matrix,classification_report,accuracy_score
from sklearn.preprocessing import OneHotEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.compose import ColumnTransformer
```

```python
from sklearn.preprocessing import LabelEncoder
len=LabelEncoder()
y=len.fit_transform(y)
```

```python
#spiltting the y and x values to train and test
x_test,x_train,y_test,y_train=train_test_split(x,y, test_size=0.2,random_state=42,stratify=y)
```

```python
data_model.columns
```

```python
#identifying numericl and categorical columns
numerical=["Total score"]
categorical=["Gender","Age range","Hear_About_us","Track","Aim","Learning_Experience","Hours_commit","Skill level","Test_Completion","Country"]
```

```python
#performing standardrization to numerical values and one hot encoding to categorical columns
preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numerical),
        ("cat",OneHotEncoder(handle_unknown="ignore"), categorical)
    ],
    remainder="drop"
)
```

```python
#fitting the logisitic regression model
pipeline=Pipeline([
    ("prep",preprocessor),
    ("clf",LogisticRegression(class_weight="balanced",max_iter=1000))
])
```

```python
pipeline.fit(x_train,y_train)
```

```python
y_pred=pipeline.predict(x_test)
```

```python
print(classification_report(y_test,y_pred))
```

```python
import joblib
joblib.dump(pipeline,"logmodel.pkl")
```

```python
#traing with randomforest classifier
from sklearn.ensemble import RandomForestClassifier
rfc=Pipeline([
    ("prep",preprocessor),
    ("classifier",RandomForestClassifier(n_estimators=100,class_weight="balanced"))
])
```

```python
rfc.fit(x_train,y_train)
```

```python
y_prediction=rfc.predict(x_test)
```

```python
print(classification_report(y_test,y_prediction))
```

```python
rfc.score(x_test,y_test)*100
```

```python
joblib.dump(rfc,"RandomClass.pkl")
```
